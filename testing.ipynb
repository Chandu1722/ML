```python
import pandas as pd
import numpy as np
```

```python
data=pd.read_csv('/content/swiggy_cleaned.csv')
data
```

```python
data.columns
```

```python
data=data.drop(columns=['rider_id','age','restaurant_latitude','city_name','restaurant_longitude', 'delivery_latitude', 'delivery_longitude','order_date','type_of_order', 'order_day_of_week'
,'order_time_hour', 'order_time_of_day'])
```

```python
num_cols=data.select_dtypes(['float64','int64']).columns
for col in num_cols:
  data[col]=data[col].fillna(data[col].mean())
```

```python
cat_cols=data.select_dtypes(['object']).columns
for col in cat_cols:
  data[col]=data[col].fillna(data[col].mode()[0])
```

```python
data.info()
```

```python
nominal_cols = ['type_of_vehicle','festival','city_type','order_day','is_weekend']
ordinal_cols = ['weather','traffic']
numeric_cols = ['ratings','vehicle_condition','multiple_deliveries','pickup_time_minutes','distance']
```

```python
data['weather'].unique()
```

```python
seq_order=[['sunny','cloudy','windy', 'fog','sandstorms','stormy'],['low','medium','high','jam']]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
```

```python
Transformer = ColumnTransformer([('Nominal Encodding',OneHotEncoder(),nominal_cols),
                ("Ordinal Encoding",OrdinalEncoder(categories=seq_order),ordinal_cols),
                ("Feature Scaling",StandardScaler(),numeric_cols)])
```

```python
X=data.drop('time_taken',axis=1)
y=data['time_taken']
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,train_size=0.2,random_state=42)
```

```python
model=Pipeline([('Transformer',Transformer),('Linear Regression',LinearRegression())])
model.fit(x_train,y_train)
y_pred_lr=model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,r2_score
mae=mean_absolute_error(y_test,y_pred_lr)
r2=r2_score(y_test,y_pred_lr)
```

```python
print(mae)
print(r2)
```

```python
from sklearn.tree import DecisionTreeRegressor
model_dt=Pipeline([('Transformer',Transformer),('Decision Tree',DecisionTreeRegressor())])
model_dt.fit(x_train,y_train)
y_pred_dt=model_dt.predict(x_test)
```

```python
mae=mean_absolute_error(y_test,y_pred_dt)
r2=r2_score(y_test,y_pred_dt)
print(mae)
print(r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
model_rf=Pipeline([('Transformer',Transformer),('Random Forest',RandomForestRegressor())])
model_rf.fit(x_train,y_train)
y_pred_rf=model_rf.predict(x_test)
```

```python
mae=mean_absolute_error(y_test,y_pred_rf)
r2=r2_score(y_test,y_pred_rf)
print(mae)
print(r2)
```
