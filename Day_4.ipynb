```python
m=data['age'].mean()
data['age']=data['age'].fillna(m)
```

```python
mo=data['weather'].mode()
```

```python
data['weather']=data['weather'].fillna('fog')
data['weather'].isna().sum()
```

```python
num_cols=data.select_dtypes(['float64','int64']).columns
```

```python
for col in num_cols:
  data[col]=data[col].fillna(data[col].mean())
```

```python
cat_cols=data.select_dtypes(['object']).columns
```

```python
for col in cat_cols:
  data[col]=data[col].fillna(data[col].mode()[0])
```

```python
ordinal_col=['traffic','city_type','festival']
traffic_order=['low','medium','high','jam']
city_order=['semi-urban','urban','metropolitian']
fest_order=['no','yes']
ordinal_order=[traffic_order,city_order,fest_order]
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder=OrdinalEncoder(categories=ordinal_order)
data[ordinal_col]=encoder.fit_transform(data[ordinal_col])
```

```python
data.info()
```

```python
nominal_data=['weather','type_of_order','type_of_vehicle','city_name','order_day_of_week',]
```

```python
data=pd.get_dummies(data,columns=nominal_data,drop_first=True)
```

```python
data
```

```python
X=data.drop('time_taken',axis=1)
y=data['time_taken']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,train_size=0.20,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(x_train,y_train)
y_pre=model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error,r2_score
mea=mean_absolute_error(y_test,y_pre)
r2=r2_score(y_test,y_pre)
print(mea)
print(r2)
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
x_train_scaled=scalar.fit_transform(x_train)
x_test_scaled=scalar.fit_transform(x_test)
```

```python
model_knn=KNeighborsRegressor(n_neighbors=17)
model_knn.fit(x_train_scaled,y_train)
y_pred_knn=model_knn.predict(x_test_scaled)
```

```python
mea=mean_absolute_error(y_test,y_pred_knn)
r2=r2_score(y_test,y_pred_knn)
print(mea)
print(r2)
```

```python
from sklearn.ensemble import RandomForestRegressor
model_rf=RandomForestRegressor(n_estimators=100,random_state=42)
model_rf.fit(x_train,y_train)
y_pred_rf=model_rf.predict(x_test)
```

```python
mea=mean_absolute_error(y_test,y_pred_rf)
r2=r2_score(y_test,y_pred_rf)
print(mea)
print(r2)
```

```python
from xgboost import XGBRFRegressor
model_xg=XGBRFRegressor(n_estimators=100,random_state=42)
model_xg.fit(x_train,y_train)
y_pred_xg=model_xg.predict(x_test)
```

```python
mea=mean_absolute_error(y_test,y_pred_xg)
r2=r2_score(y_test,y_pred_xg)
print(mea)
print(r2)
```
