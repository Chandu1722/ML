```python
import pandas as pd
```

```python
train=pd.read_csv('train.csv')
test=pd.read_csv('test.csv')
train.head()
```

```python
test.head()
```

```python
test.isna().sum()
```

```python
train=train.dropna()
```

```python
train
```

```python
train['gender']=train['gender'].str.strip().map({'Female':0,'Male':1})
test['gender']=test['gender'].str.strip().map({'Female':0,'Male':1})
```

```python
for col in train.columns:
  if train[col].dtype=='object':
    if len(train[col].unique())==2:
      train[col]=train[col].str.strip().map({'No':0,'Yes':1})
for col in test.columns:
  if test[col].dtype=='object':
    if len(test[col].unique())==2:
      test[col]=test[col].str.strip().map({'No':0,'Yes':1})
```

```python
train
```

```python
train['MultipleLines']=train['MultipleLines'].str.strip().map({'No':0,'No phone service':1,'Yes':2})
test['MultipleLines']=test['MultipleLines'].str.strip().map({'No':0,'No phone service':1,'Yes':2})
```

```python
train['InternetService'].unique()
```

```python
train['InternetService']=train['InternetService'].str.strip().map({'Fiber optic':0,'DSL':1,'No':2})
test['InternetService']=test['InternetService'].str.strip().map({'Fiber optic':0,'DSL':1,'No':2})
```

```python
for col in train.columns:
  if train[col].dtype=='object':
    print(col,train[col].unique())
for col in test.columns:
  if test[col].dtype=='object':
    print(col,test[col].unique())
```

```python
train['OnlineSecurity']=train['OnlineSecurity'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
train['OnlineBackup']=train['OnlineBackup'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
train['DeviceProtection']=train['DeviceProtection'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
train['TechSupport']=train['TechSupport'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
train['StreamingTV']=train['StreamingTV'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
train['StreamingMovies']=train['StreamingMovies'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
train['Contract']=train['Contract'].str.strip().map({'Month-to-month':0,'One year':1,'Two year':2})
train['PaymentMethod']=train['PaymentMethod'].str.strip().map({'Electronic check':0,'Mailed check':1,'Bank transfer (automatic)':2,'Credit card (automatic)':3})
test['OnlineSecurity']=test['OnlineSecurity'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
test['OnlineBackup']=test['OnlineBackup'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
test['DeviceProtection']=test['DeviceProtection'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
test['TechSupport']=test['TechSupport'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
test['StreamingTV']=test['StreamingTV'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
test['StreamingMovies']=test['StreamingMovies'].str.strip().map({'No':0,'No internet service':1,'Yes':2})
test['Contract']=test['Contract'].str.strip().map({'Month-to-month':0,'One year':1,'Two year':2})
test['PaymentMethod']=test['PaymentMethod'].str.strip().map({'Electronic check':0,'Mailed check':1,'Bank transfer (automatic)':2,'Credit card (automatic)':3})
```

```python
train
```

```python
corr=train.corr()['Churn'].abs().sort_values(ascending=False)
print(corr)
```

```python
from sklearn.model_selection import train_test_split
X=train.drop(columns=['id','Churn'])
y=train['Churn']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
y_full=test.drop(columns=['id'])
```

```python
from sklearn.metrics import roc_auc_score
```

```python
from sklearn.ensemble import RandomForestClassifier
modelRF=RandomForestClassifier()
modelRF.fit(X_train,y_train)
y_pred_RF=modelRF.predict_proba(X_test)[:,1]
aucRF=roc_auc_score(y_test,y_pred_RF)
print(aucRF)
```

```python

Y_pred_full=modelRF.predict_proba(y_full)[:,1]
sub=pd.DataFrame({'id':test['id'],'Churn':Y_pred_full})
sub.to_csv('SubmissionRF.csv',index=False)
```

```python
from xgboost import XGBClassifier
modelXGB=XGBClassifier()
modelXGB.fit(X_train,y_train)
y_pred_XGB=modelXGB.predict_proba(y_full)[:,1]
subm=pd.DataFrame({'id':test['id'],'Churn':y_pred_XGB})
subm.to_csv('SubmissionXGB.csv',index=False)
```
