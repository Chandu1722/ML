```python
import os
import shutil
import random
```

```python
source_dir = "/content/dataset/animals"
destination_dir = "/content/data_split"

categories = ['cat', 'dog']

# Create folders
for split in ['train', 'test']:
    for category in categories:
        os.makedirs(os.path.join(destination_dir, split, category), exist_ok=True)

# Split ratio
split_ratio = 0.8  # 80% train, 20% test

for category in categories:
    folder_path = os.path.join(source_dir, category)
    images = os.listdir(folder_path)
    random.shuffle(images)

    split_point = int(len(images) * split_ratio)

    train_imgs = images[:split_point]
    test_imgs = images[split_point:]

    # Copy files to train
    for img in train_imgs:
        shutil.copy(os.path.join(folder_path, img),
                    os.path.join(destination_dir, 'train', category, img))

    # Copy files to test
    for img in test_imgs:
        shutil.copy(os.path.join(folder_path, img),
                    os.path.join(destination_dir, 'test', category, img))

print("✅ Dataset successfully split into train and test!")
```

```python
import tensorflow as tf

img_size = (150, 150)
batch_size = 32

train_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "/content/data_split/train",
    image_size=img_size,
    batch_size=batch_size,
)

test_ds = tf.keras.preprocessing.image_dataset_from_directory(
    "/content/data_split/test",
    image_size=img_size,
    batch_size=batch_size,
)
```

```python
from tensorflow.keras import models,layers
model=models.Sequential([
    layers.Conv2D(32,(3,3),activation='relu',input_shape=(150,150,3)),
    layers.MaxPooling2D(),
    layers.Conv2D(64,(3,3),activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dense(64,activation='relu'),
    layers.Dense(1,activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
history=model.fit(train_ds,validation_data=test_ds,epochs=10)
```
